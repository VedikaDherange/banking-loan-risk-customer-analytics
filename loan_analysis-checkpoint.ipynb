```python
#loan amount distribution
plt.figure(figsize=(8,5))
plt.hist(df['loan_amnt'],bins=20)
plt.xlabel("Loan Amount")
plt.ylabel("Frequency")
plt.title("Loan Amount Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['person_income'],bins=20)
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.title("Income Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['credit_score'],bins=20)
plt.xlabel("Credit Score")
plt.ylabel("Frequency")
plt.title("Credit Score Distribution")
plt.show()
```

```python
loan_purpose=df["loan_intent"].value_counts()
plt.figure(figsize=(8,5))
plt.bar(loan_purpose.index,loan_purpose.values)
plt.xticks(rotation=45)
plt.xlabel("Loan Purpose")
plt.ylabel("Count")
plt.title("Loan Purpose Analysis")
plt.show()
```

```python
defaults=df["previous_loan_defaults_on_file"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(
    defaults.values,
    labels=defaults.index,
    autopct="%1.1f%%"
)
plt.title("Previous Loan Default Analysis")
plt.show()
```
